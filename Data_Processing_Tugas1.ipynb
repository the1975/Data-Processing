1. Buatlah program pyhton dimana bisa perhitungan otomatis dengan python untuk mencari

*   Volume Tabung
*   Volume balok
*   Luas Kerucut

dimana nilai nilai input berasal dari user, berikut contohnya

```bash
Masukan Panjangnya(cm) = _2_
Masukan Lebarnya(cm) = _3_
Masukan Tingginya(cm) = _1_
Maka Volume balok tersebut adalah 6
```

````python
## Tulis Kode Disini

p = float(input("Masukan Panjangnya(cm) = "))
l = float(input("Masukan Lebarnya(cm) = "))
t = float(input("Masukan Tingginya(cm) = "))
r = float(input('Masukan Jari-jarinya(cm) ='))
s = float(input('Masukan garis pelukis(cm) ='))
pi = 3.14

volume_balok = p * l * t
volume_tabung = pi * r * r * t
luas_kerucut = pi * r * (r + s)

output = f'''
Maka Volume balok tersebut adalah, {volume_balok} cm
Maka Volume tabung tersebut adalah, {volume_tabung} cm
Maka Luas kerucut tersebut adalah, {luas_kerucut} cm
'''

print(output)
````

2. Buatlah program untuk mencari diskon dari suatu pembelian barang dengan kondisi
*   Jika Pembelian diatas 100.000 dapat diskon 10%
*   Jika Pembelian diatas 50.000 diskon 5%
*   Selain itu tidak dapat diskon

contoh

```bash
Masukan Total Pembelian = _200000_
Diskon Anda = 20000
Total Bayar Anda = 180000
```

````python
## Tulis Kode Disini
beli = float(input("Masukkan Total Pembelian = "))
if beli >= 100.000:
  diskon = beli * 0.10
elif beli >= 50.000:
  diskon = beli * 0.05
else:
  diskon = 0
total = beli - diskon
print ("Diskon anda =", diskon)
print ("Total Bayar Anda", total)
````
