## Looping

1. Dengan menggunakan perulangan WHILE buatlah deret angka dari 10 s/d 1

````python
y = 11
i = 0

while i < 10:
  i += 1
  y -= 1
  print(y)
````

3. Buatlah program untuk mencari jumlah dari suatu deret angka bulat dimana
batasnya ditentukan oleh user, berikut adalah contoh tampilannya

```bash
Masukan Batas = _5_
Jumlah deret angka 1 s/d 5 adalah 15
```

````python
batas = int(input('Masukkan batas = '))
jumlah = 0

for i in range(1, batas +1):
    jumlah += i
    print(jumlah)
````

2. Buatlah program untuk mencatat data Task atau Tugas dimana tugas tersebut bisa ditambahkan dan di hapus jika sudah dikerjakan. Berikut adalah contoh tampilannya

```bash
To Do Apps
==========

Masukan Tugas _Tugas 3_

Daftar Tugas
Tugas 1
Tugas 2
Tugas 3

Hapus Tugas indeks ke - _2_

Daftar Tugas
Tugas 1
Tugas 2
```

````python
# List untuk menyimpan tugas
tugas = []

print("To Do Apps")
print("-----------")

# Tambah tugas
n = int(input("Berapa tugas yang ingin dimasukkan? "))

for i in range(n):
    t = input(f"Masukkan Tugas {i+1}: ")
    tugas.append(t)

# Tampilkan tugas
print("\nDaftar Tugas")
for i in range(len(tugas)):
    print(f"Tugas {i+1}: {tugas[i]}")

# Hapus tugas berdasarkan index
hapus = int(input("\nHapus tugas indeks ke (mulai dari 1): "))

# karena index python mulai dari 0
tugas.pop(hapus - 1)

# Tampilkan lagi
print("\nDaftar Tugas setelah dihapus")
for i in range(len(tugas)):
    print(f"Tugas {i+1}: {tugas[i]}")
````

1. Buatlah Aplikasi Penjualan Barang Sederhana dimana aplikasi tersebut bisa menambahkan barang beserta harga dan jumlah beli, dan nantinya akan mengetahui total belanja keseluruhan dan bisa menambahkan pembayaran untuk diketahui apakah ada kembalian atau tidak, berikut adalah contoh tampilannya

```bash
+-----------------------+
+ Aplikasi Kasir Python +
+-----------------------+

Nama Barang = __Pencil__
Harga = __1000__
jumlah = __2__
Subtotal = 2000

Input Lagi (Ya/Tidak) = __Ya__

Nama Barang = __Buku__
Harga = __3000__
jumlah = __1__
Subtotal = 3000

Input Lagi (Ya/Tidak) = __Tidak__

Total Keseluruhan = 5000
Pembayaran = __10000__
Kembali = 5000

```

````python
print("+------------------------+")
print("+ Aplikasi Kasir Python +")
print("+------------------------+")

total = 0
hargaPensil = 2000
hargaBuku = 3000
hargaPengapus = 1000
hargaComic = 5000

while True:
    nama = int(input("Selamat datang, silahkan pilih: \n1. Pensil \n2. Buku \n3. Penghapus \n4. Comic jjk\n \nPilihan = "))
    jumlah = int(input("Jumlah = "))

    if nama == 1:
        harga = hargaPensil
    elif nama == 2:
          harga = hargaBuku
    elif nama == 3:
          harga = hargaPengapus
    elif nama == 4:
          harga = hargaComic
    else:
        print('tolong pilih sesuai etalase')
        continue
    subtotal = harga * jumlah
    print("Subtotal =", subtotal)

    total += subtotal

    lagi = input("Mau pesan lagi? (Y/N) = ").lower()

    if lagi == "n":
        break

print("\nTotal Keseluruhan =", total)

while True:
      bayar = int(input("Pembayaran = "))

      if bayar < total:
          total = total - bayar
          print("Maaf uang anda kurang, tolong bayar kekuarannya")
          print("\nTotal kekuarannya =", total)

      else:
          kembali = bayar - total
          break

print("Kembali =", kembali)
````
