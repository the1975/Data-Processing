perhitungan otomatis dengan python untuk mencari
- Volume Tabung
- Volume Balok

```bash
Masukan Panjangnya(cm) = _2_
```

````python
def tabung(r, t):
    return 3.14 * r * r * t
````

````python
def balok(p, l, t):
    return p * l * t
````

````python
def kerucut(r, t):
    return 1/3 * 3.14 * r * r * t
````

````python
r = float(input("Masukkan jari-jari: "))
t = float(input("Masukkan tinggi: "))
l = float(input("Masukkan lebar: "))
p = float(input("Masukkan panjang: "))

print(f"\nVolume Tabung adalah {tabung(r, t)}")
print(f"Volume Balok adalah {balok(p, l, t)}")
print(f"Volume Kerucut adalah {kerucut(r, t)}")
````

2. Dengan menggunakan Fungsi, Buatlah program untuk **mencatat data Task** atau Tugas dimana tugas tersebut bisa **ditambahkan** dan **dihapus** jika sudah dikerjakan.

Berikut adalah contoh tampilannya

```bash
ToDo Apps
1. Tampilkan Task
2. Tambah Task
3. Hapus Task
4. Keluar
Pilihan Anda [1 s/d 4] = __
```

````python
tasks = []

def tampilkan_task():
    if not tasks:
        print("Belum ada task.")
    else:
        print("\nDaftar Task:")
        for i, task in enumerate(tasks, start=1):
            print(f"{i}. {task}")

def tambah_task():
    task = input("Masukkan task baru: ")
    tasks.append(task)
    print("Task berhasil ditambahkan!")

def hapus_task():
    tampilkan_task()
    if tasks:
        try:
            nomor = int(input("Masukkan nomor task yang mau dihapus: "))
            if 1 <= nomor <= len(tasks):
                removed = tasks.pop(nomor - 1)
                print(f"Task '{removed}' berhasil dihapus!")
            else:
                print("Nomor tidak valid!")
        except ValueError:
            print("Input harus angka!")

def menu():
    while True:
        print("\nToDo Apps")
        print("1. Tampilkan Task\n2. Tambah Task\n3. Hapus Task\n4. Keluar")

        pilihan = int(input("Pilihan Anda [1 s/d 4] = "))

        if pilihan == 1:
            tampilkan_task()
        elif pilihan == 2:
            tambah_task()
        elif pilihan == 3:
            hapus_task()
        elif pilihan == 4:
            print("Keluar dari program...")
            break
        else:
            print("Pilihan tidak valid!")

# Jalankan program
menu()
````

a. **Buat Function** untuk mencari

b. **Buat Class Produk** dengan Atribut:

- nama
- harga

dengan method atau function sebagai berikut kategori():
- Mahal (> 2000)
- Sedang (500–2000)
- Murah (< 500)

Output yang diharapkan adalah

```bash
Total harga: xxx
Rata-rata: xxx
Produk termahal: xxx

Laptop - Mahal
Mouse - Murah
dst
```


````python
data =[7000, 150, 300, 2500, 1200]

def  rata(data):
  return sum(data)/len(data)

def jumlah(data):
  return sum(data)

def PalingMahal(data):
  return max(data)
````

````python
class barang:
  def __init__(self, nama, harga):
    self.nama = nama
    self.harga = harga

    if harga >= 2000:
      self.status = 'Mahal'
    elif harga >= 500:
      self.status = 'Sedang'
    else:
      self.status = 'Murah'


brg_list = [
    barang('Laptop', 7000),
    barang('Keyboard', 150),
    barang('Mouse', 300),
    barang('Monitor', 2500),
    barang('Printer', 1200),
]
````

````python
print(f'Total harga: {jumlah(data)}')
print(f'Rata-rata: {rata(data)}')
print(f'Produk termahal: {PalingMahal(data)}\n')

for n in brg_list:
  print(f'{n.nama} : {n.status}')
````
