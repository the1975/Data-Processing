**Data Cleaning**

```python
# Handle mising value pada kolom overview
df['overview'] = df['overview'].fillna('No overview Available')

# Menganti tipe data pada kolom release_date
df['release_date'] = pd.to_datetime(df['release_date'])

df['year'] = df['release_date'].dt.year
df['month'] = df['release_date'].dt.month
df['day'] = df['release_date'].dt.day

# Mengapus kolom unnamed: 0
df.drop(columns=['Unnamed: 0'], inplace=True)

# Menghapus data duplikat
df.drop_duplicates(inplace=True)
```

```python
q25 = df['popularity'].quantile(0.25)
q75 = df['popularity'].quantile(0.75)

df['popularity_category'] = pd.cut(df['popularity'], bins = [0, q25, q75, float('inf')], labels = ['Kurang populer', 'Sedang', 'Populer'])
```
