```python
# install library buat scrapping
!pip install beautifulsoup4
!pip install requests
!pip install pandas

# import library yang dibutuhkan
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
```

```python
# buat header
Headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/124.0.0.0 Safari/537.36"
    ),
    'Accept-Language': 'en-US,en;q=0.5'
}

# base url yang mau di scrapping
base_url = 'https://referensi.data.kemendikdasmen.go.id/residu/pesertadidik/wilayah/051400/2'

print('persiapan selesai...')
```

```python
# tes koneksi
res = requests.get(base_url, headers=Headers, timeout=10)
print(res.status_code)
```

```python
# parse html
soup = BeautifulSoup(res.text, 'html.parser')
print(soup.prettify()[:6000])
```

```python
# Coba beberapa selector umum Freelancer
candidates = [
    "#example tbody tr",
]

for selector in candidates:
    found = soup.select(selector)
    print(f"'{selector}' → {len(found)} elemen ditemukan")
```

```python
rows = soup.select("#example tbody tr")

data = []

for row in rows:
  cols = row.find_all('td')

  nomor = cols[0].text.strip()
  nama_kecamatan = cols[1].text.strip()
  jumlah_peserta_didik = cols[2].text.strip()
  residu_nisn = cols[3].text.strip()
  residu_kependudukan = cols[4].text.strip()
  residu_wilayah = cols[5].text.strip()

  data.append({
      'nomor': nomor,
      'nama_kecamatan': nama_kecamatan,
      'jumlah_peserta_didik': jumlah_peserta_didik,
      'residu_nisn': residu_nisn,
      'residu_kependudukan': residu_kependudukan,
      'residu_wilayah': residu_wilayah
  })

df = pd.DataFrame(data)
df.head()
```
