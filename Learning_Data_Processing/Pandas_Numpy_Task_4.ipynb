## EXPLORATORY DATA ANALYSIS (EDA)

lakukan EDA pada tiap dataset yang digunakan

1. Statistik Deskriptif

```python
# kode disini
print("Statistik Deskriptif Titanic\n")
display(df_titanic.describe(include='all'))

print("\nStatistik Deskriptif Students Performance\n")
display(df_students.describe(include='all'))

print("\nStatistik Deskriptif Superstore\n")
display(df_superstore.describe(include='all'))
```

2. Missing Value

```python
datasets = [('Titanic', df_titanic), ('Students Performance', df_students), ('Superstore', df_superstore)]

for name, df in datasets:
    print(f"Missing Values: {name}\n")
    missing = df.isnull().sum()
    print(missing[missing > 0])
    print(f"Total sel kosong: {df.isnull().sum().sum()}\n")
```

**Analisis tiap dataset**

1. Apakah ada data kosong?
2. Kolom mana yang bermasalah?

**Dataset Titanic**: Ada data kosong pada kolom Age (177), Cabin (687), dan Embarked (2).

**Dataset Students Performance**: Tidak ada data kosong.

**Dataset Superstore**: Tidak ada data kosong.Jawaban Anda disini

3. Analisis Numerik

- Rerata
- Nilai tertinggi
- nilai terendah

```python
# Memilih beberapa kolom numerik penting dari tiap dataset untuk dianalisis
print("Analisis Numerik Titanic (Age & Fare)\n")
display(df_titanic[['Age', 'Fare']].agg(['mean', 'max', 'min']))

print("\nAnalisis Numerik Students Performance (Scores)\n")
display(df_students[['math score', 'reading score', 'writing score']].agg(['mean', 'max', 'min']))

print("\nAnalisis Numerik Superstore (Sales & Profit)\n")
display(df_superstore[['Sales', 'Profit']].agg(['mean', 'max', 'min']))
```

4. Filtering Data

- Lakukan filtering dan analisis data yang menarik dari masing-masing dataset yang digunakan

```python
print("Titanic Dataset\n")
titanic_filter = df_titanic[(df_titanic['Sex'] == 'female') & (df_titanic['Pclass'] == 1) & (df_titanic['Survived'] == 1)]
print(f"Jumlah penumpang wanita kelas 1 yang selamat: {len(titanic_filter)} orang")
print(garis)
print("StudentPerfomance Dataset\n")
student_filter = df_students[df_students['math score'] == 100]
print(f"Jumlah siswa dengan nilai Matematika sempurna (100): {len(student_filter)} orang")
print("Rata-rata reading score mereka:", student_filter['reading score'].mean())
print(garis)
print("SampleSuperstore Dataset\n")
superstore_filter = df_superstore[(df_superstore['Profit'] < 0) & (df_superstore['Sales'] > 1000)]
print(f"Jumlah transaksi rugi meskipun penjualan di atas $1000: {len(superstore_filter)} transaksi")
print(garis)
```
