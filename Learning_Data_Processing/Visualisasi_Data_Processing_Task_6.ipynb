**5. data cleaning**

> titanic dataset

```python
# ubah mising value dataset titanic kolom cabin dan embaked dengan unknown
df_titanic['Cabin'] = df_titanic['Cabin'].fillna('Unknown')
df_titanic['Embarked'] = df_titanic['Embarked'].fillna('Unknown')

# ubah mising value dataset titanic kolom age dengan median
df_titanic['Age'] = df_titanic['Age'].fillna(df_titanic['Age'].median())
```

```python
# cek mising value dataset titanic
df_titanic.isnull().sum().sort_values(ascending=False)
```

> student performance dataset

```python
# tambah rata rata nilai dataset student performance
df_student['rata-rata'] = (df_student['math score'] + df_student['reading score'] + df_student['writing score']) / 3

# tambah status lulus/gagal kalo nilai kurang dari 75
df_student['status'] = df_student['rata-rata'].apply(lambda x: 'Lulus' if x >= 75 else 'Gagal')
```

```python
# cek data
df_student.head(10)
```

> sample superstore dataset

```python
# ubah kolom order date dan ship date pada dataset superstore menjadi datetime
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
df_superstore['Ship Date'] = pd.to_datetime(df_superstore['Ship Date'])

# buat fitur waktu pada dataset superstore
df_superstore['Order Year'] = df_superstore['Order Date'].dt.year
df_superstore['Order Month'] = df_superstore['Order Date'].dt.month
df_superstore['Order Day'] = df_superstore['Order Date'].dt.day

# grouping penjualan per bulan
monthly_sales = df_superstore.groupby('Order Month')['Sales'].sum()

# buat lama pengiriman pada dataset superstore
df_superstore['Pengiriman Lama (hari)'] = (df_superstore['Ship Date'] - df_superstore['Order Date']).dt.days
```

## Data Visualization

lakukan VISUALISASI DATA pada tiap dataset dengan menggunakan

1. Bar Chart


```python
# titanic dataset bar chart
sns.countplot(
    x='Sex',              # sumbu x menunjukkan gender
    hue='Survived',
    data=df_titanic,
    palette=['red', 'green']
)

plt.title('Survival Berdasarkan Gender')
plt.ylabel('Jumlah penumpang')
plt.xlabel('Gender penumpang')

plt.show()
```

**Penjelasan :**
> grafik ini menunjukkan jumlah penumpang titanic berdasarkan gender dan status keselamatan, warna merah menunjukka penumpang yang tidak selamat dan warma hijau menunjukkan penumpang yang selamat.

**Insight :**
> berdasarkan grafik penumpang perempuan memiliki keselamatan yang lebih tinggi daripada penumpang laki-laki. sebaliknya, sebaliknya sebagian besar penumpang laki-laki tidak berhasil selamat dalam kecelakaan kapal titanic

```python
# bar chart dataset student performance
plt.figure(figsize=(8,5))

sns.barplot(
    x='gender',
    y='rata-rata',
    data=df_student,
    palette=['pink', 'skyblue']
)

plt.title('Rata-rata Nilai Berdasarkan Gender')
plt.xlabel('Gender')
plt.ylabel('Rata-rata Nilai')

plt.show()
```

**Penjelasan :**
> grafik menunjukkan nilai rata-rata siswa berdasarkan gender. nilai rata-rata ini diperoleh dari gabungan nilai matematika, membaca, dan menulis

**Insight :**   
> berdasarkan grafik, nilai rata-rata siswa perempuan sedikit lebih tinggi dibandingkan siswa laki-laki. namun, perbedaan tidak terlalu signifikan, sehimgga performa akademik kedua gender cenderung relatif seimbang

```python
# bar chart dataset superstore
plt.figure(figsize=(8,5))

sns.barplot(
    x='Category',
    y='Profit',
    data=df_superstore,
    palette=['brown', 'royalblue', 'teal']
)

plt.title('Profit Berdasarkan Category')
plt.xlabel('Category')
plt.ylabel(' Jumlah Profit')

plt.show()
```

**Penjelasan :**
> grafik ini menunjukkan perbandingan profit pada setiap kategori produk di dataset superstore.

**Insight :**   
> kategori teknologi menghasilkan profit yang jauh lebih tinggi dibanding kategori lainnya. perbedaan profit kategori teknologi sangat signifikan, dimana bahkan lebih dari dua kali lipat kategori lainnya

2. Pie Chart


```python
# pie chart titanic dataset
survival_counts = df_titanic['Survived'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    survival_counts,
    labels=['Tidak Selamat', 'Selamat'],
    autopct='%1.1f%%',
    colors=['salmon', 'mediumseagreen']
)

plt.title('Persentase Survival Penumpang')

plt.show()
```

**Penjelasan :**
> Pie chart menunjukkan persentase penumpang yang selamat dan tidak selamat pada tragedi Titanic.

**Insight :**   
> Sebagian besar penumpang Titanic tidak berhasil selamat dibanding penumpang yang selamat.

```python
# Pie chart dataset student performance
plt.figure(figsize=(6,6))

plt.pie(
    df_student['status'].value_counts(),
    labels=['Gagal', 'Lulus'],
    autopct='%1.1f%%',
    colors=['gold', 'deepskyblue']
)

plt.title('Persentase Kelulusan Siswa')

plt.show()
```

**Penjelasan :**
> Pie chart menunjukkan persentase siswa yang lulus dan gagal berdasarkan rata-rata nilai. Status lulus ditentukan jika rata-rata nilai siswa lebih besar atau sama dengan 75.

**Insight :**   
> Berdasarkan grafik, sebagian besar siswa tidak memenuhi batas kelulusan yang telah ditentukan. Sebanyak 67.6% siswa dinyatakan gagal, sedangkan hanya 32.4% siswa yang lulus.
Hal ini menunjukkan bahwa standar nilai 75 cukup tinggi untuk sebagian besar siswa pada dataset tersebut.

```python
# Pie Chart dataset superstore
segment_counts = df_superstore['Segment'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    segment_counts,
    labels=segment_counts.index,
    autopct='%1.1f%%',
    colors=['skyblue', 'blue', 'orange']
)

plt.title('Persentase Segment Customer')

plt.show()
```

**Penjelasan :**
> Pie chart menunjukkan distribusi segment customer pada dataset Superstore.

**Insight :**   
> Berdasarkan pie chart, segment Consumer mendominasi jumlah customer dibanding segment lainnya. Hal ini menunjukkan bahwa sebagian besar transaksi pada Superstore berasal dari pelanggan individu atau konsumen umum.

3. Histogram


```python
# Histogram titanic dataset
plt.figure(figsize=(8,5))

sns.histplot(
    data=df_titanic,
    x='Age',
    bins=20,
    kde=True,
    color='coral'
)

plt.title('Distribusi Umur Penumpang')
plt.xlabel('Umur')
plt.ylabel('Jumlah')

plt.show()
```

**Penjelasan :**
> Histogram menunjukkan distribusi umur penumpang Titanic berdasarkan jumlah penumpang pada setiap rentang usia.

**Insight :**   
> Sebagian besar penumpang berada pada rentang usia dewasa muda. Sementara itu, jumlah penumpang dengan usia lanjut terlihat lebih sedikit dibanding kelompok usia lainnya.

```python
# Histogram student performance dataset
plt.figure(figsize=(8,5))

sns.histplot(
    data=df_student,
    x='rata-rata',
    bins=20,
    kde=True,
    color='royalblue'
)

plt.title('Distribusi Nilai Rata-Rata')
plt.xlabel('Nilai Rata-Rata')
plt.ylabel('Jumlah')

plt.show()
```

**Penjelasan :**
> Histogram menunjukkan distribusi rata-rata nilai siswa berdasarkan gabungan nilai matematika, membaca, dan menulis.

**Insight :**   
> Sebagian besar siswa memiliki rata-rata nilai pada rentang menengah. Hanya sedikit siswa yang memperoleh nilai sangat rendah maupun sangat tinggi, sehingga distribusi nilai terlihat cukup terpusat pada nilai tengah rata-rata.

```python
# Histogram superstore dataset
plt.figure(figsize=(8,5))

sns.histplot(
    df_superstore['Profit'],
    bins=30,
    kde=True,
    color='limegreen'
)

plt.title('Distribusi Profit')
plt.xlabel('Profit')
plt.ylabel('Jumlah Transaksi')

plt.show()
```

**Penjelasan :**

**Insight :**   
> Berdasarkan histogram, sebagian besar transaksi menghasilkan profit yang berada di sekitar angka 0 atau profit kecil. Hal ini menunjukkan bahwa mayoritas transaksi hanya memberikan keuntungan dalam jumlah rendah, sedangkan transaksi dengan profit sangat tinggi maupun kerugian besar jumlahnya lebih sedikit.
